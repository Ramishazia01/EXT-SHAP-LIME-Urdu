```python
# ============================================================
# PLOTS FOR ARABIC DATA 
# ============================================================

import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# RTL library imports
from shap_rtl.plots import heatmap, scatter
from shap_rtl import RTLSHAPExplainer

# ============================================================
# CELL 1: LOAD ARABIC DATA
# ============================================================

print("="*80)
print(" LOADING ARABIC DATASET")
print("="*80)

# Load Arabic data
file_path = r'C:\Users\USER\OneDrive - Higher Education Commission\Dr. shahid project\SHAP_package\Datasets\L-HSAB.txt'

# Load dataset
try:
    df = pd.read_csv(file_path, sep='\t', header=None, names=['text', 'label'])
except:
    try:
        df = pd.read_csv(file_path, sep=',', header=None, names=['text', 'label'])
    except:
        df = pd.read_csv(file_path, sep='\s+', header=None, names=['text', 'label'])

# Clean data
df = df.dropna()
df['text'] = df['text'].astype(str).str.strip()
df['label'] = df['label'].astype(str).str.strip()

# Map labels to binary
unique_labels = df['label'].unique()
label_map = {}
normal_labels = ['normal', '0', 'non-abusive', 'non abusive', 'not abusive']
abusive_labels = ['abusive', '1', 'hate', 'offensive', 'hate speech', 'toxic']

for label in unique_labels:
    label_lower = str(label).lower().strip()
    if label_lower in normal_labels:
        label_map[label] = 0
    elif label_lower in abusive_labels:
        label_map[label] = 1
    else:
        try:
            num_label = float(label)
            label_map[label] = 0 if num_label == 0 else 1
        except:
            if any(word in label_lower for word in ['abuse', 'hate', 'offens', 'toxic']):
                label_map[label] = 1
            else:
                label_map[label] = 0

df['label'] = df['label'].map(label_map)
df = df.dropna(subset=['label'])
df['label'] = df['label'].astype(int)

class_names = {0: 'Normal', 1: 'Abusive'}

print(f" Dataset: {len(df)} samples")
for class_id, name in class_names.items():
    count = (df['label'] == class_id).sum()
    print(f"  {name}: {count} ({count/len(df)*100:.1f}%)")

# Target sample - NORMAL
target_sample = "يسعد صباحك معالي الوزير"
print(f"\n Target Sample: {target_sample}")
print(f"   Label: Normal")

# ============================================================
# CELL 2: TRAIN MODEL
# ============================================================

print("\n" + "="*80)
print("🔧 TRAINING MODEL")
print("="*80)

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label'], 
    test_size=0.2, 
    random_state=42,
    stratify=df['label']
)

# Vectorizer with word analyzer for Arabic
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    analyzer='word',
    max_features=5000,
    min_df=3,
    max_df=0.90,
    sublinear_tf=True
)

X_train_vec = vectorizer.fit_transform(X_train).toarray()
X_test_vec = vectorizer.transform(X_test).toarray()

# Train model
model = LogisticRegression(
    max_iter=1000, 
    class_weight='balanced', 
    random_state=42, 
    C=1.0,
    solver='liblinear'
).fit(X_train_vec, y_train)

print(f" Model trained!")
print(f"Train Accuracy: {model.score(X_train_vec, y_train):.4f}")
print(f"Test Accuracy: {model.score(X_test_vec, y_test):.4f}")

feature_names = list(vectorizer.get_feature_names_out())
print(f"Total features: {len(feature_names)}")

# ============================================================
# CELL 3: SETUP RTL EXPLAINER
# ============================================================

print("\n" + "="*80)
print(" SETUP RTL EXPLAINER")
print("="*80)

X_background = X_train_vec[:100]
masker = shap.maskers.Independent(data=X_background, max_samples=100)
shap_explainer = shap.LinearExplainer(model, masker=masker)

# Create RTL explainer
explainer_rtl = RTLSHAPExplainer(
    language="Arabic",
    model=model,
    vectorizer=vectorizer,
    api_type="groq"
)

print(" RTL SHAP explainer ready!")

# ============================================================
# CELL 4: ANALYZE TARGET SAMPLE - TOP 5 FEATURES
# ============================================================

print("\n" + "="*80)
print(" ANALYZING TARGET SAMPLE - TOP 5 FEATURES")
print("="*80)

sample_text = target_sample
class_id = 0  # Normal
max_features = 5  # TOP 5 FEATURES ONLY

print(f"📝 Sample: {sample_text}")
print(f"   Label: Normal")
print(f"   Max Features: {max_features}")
print("-"*60)

# Get SHAP values for the target sample
vec = vectorizer.transform([sample_text]).toarray()[0]
present_idx = np.where(vec > 0)[0]

if len(present_idx) > 0:
    shap_values = shap_explainer(vec.reshape(1, -1))
    
    # Get present features
    present_feature_names = [feature_names[i] for i in present_idx]
    present_shap_values = shap_values.values[0][present_idx]
    present_data = vec[present_idx]
    
    # Sort by absolute SHAP value and take TOP 5
    sorted_idx = np.argsort(np.abs(present_shap_values))[::-1][:max_features]
    sorted_features = [present_feature_names[i] for i in sorted_idx]
    sorted_shap = present_shap_values[sorted_idx]
    sorted_data = present_data[sorted_idx]
    
    # Get base value
    base_value = shap_values.base_values[0]
    
    # Get prediction
    pred = model.predict(vec.reshape(1, -1))[0]
    pred_probs = model.predict_proba(vec.reshape(1, -1))[0]
    
    print(f"Model Prediction: {class_names[pred]}")
    print(f"  Normal: {pred_probs[0]:.3f}")
    print(f"  Abusive: {pred_probs[1]:.3f}")
    
    print(f"\n📊 TOP {max_features} FEATURES in Target Sample:")
    print("="*60)
    for i, (name, val) in enumerate(zip(sorted_features, sorted_shap), 1):
        direction = "→ NORMAL" if val < 0 else "→ ABUSIVE"
        emoji = "🟢" if val < 0 else "🔴"
        print(f"  {i}. {emoji} '{name}': {val:+.4f} {direction}")

# ============================================================
# CELL 5: HEATMAP PLOT - Using RTL Explainer
# ============================================================

print("\n" + "="*80)
print(" HEATMAP PLOT - Using RTL Explainer")
print("="*80)

n_samples = 80
max_features = 5  # TOP 5 FEATURES ONLY

print(f" Analyzing {n_samples} samples with top {max_features} features")
print(f" Target sample: {target_sample}")
print("-"*60)

# Get SHAP values for multiple samples
shap_values_list = []
valid_samples = []

sample_indices = np.random.choice(len(X_test_vec), min(n_samples, len(X_test_vec)), replace=False)

print("📊 Computing SHAP values...")
for idx in sample_indices:
    vec_sample = X_test_vec[idx]
    if np.sum(vec_sample) > 0:
        shap_val = shap_explainer(vec_sample.reshape(1, -1))
        shap_values_list.append(shap_val.values[0])
        valid_samples.append(vec_sample)

if shap_values_list:
    shap_values_all = np.array(shap_values_list)
    valid_samples = np.array(valid_samples)
    
    # Get top features - TOP 5
    top_features_idx = np.argsort(np.abs(shap_values_all).mean(axis=0))[::-1][:max_features]
    top_feature_names = [feature_names[i] for i in top_features_idx]
    
    print(f"\n TOP {max_features} FEATURES across dataset:")
    print("="*60)
    for i, name in enumerate(top_feature_names):
        avg_impact = np.abs(shap_values_all[:, i]).mean()
        mean_shap = np.mean(shap_values_all[:, i])
        direction = "ABUSIVE" if mean_shap > 0 else "NORMAL"
        print(f"  {i+1}. '{name}':")
        print(f"     Avg Impact: {avg_impact:.4f}")
        print(f"     Mean SHAP: {mean_shap:+.4f} → {direction}")
    
    # Check if target sample features are in top features
    print(f"\n Target sample features in top list:")
    print("-"*60)
    for feat in sorted_features:
        if feat in top_feature_names:
            print(f"   '{feat}' → Found in top features")
        else:
            print(f"   '{feat}' → Not in top features")
    
    # Create SHAP Explanation for heatmap
    shap_exp_heatmap = shap.Explanation(
        values=shap_values_all[:, top_features_idx],
        base_values=shap_explainer.expected_value if hasattr(shap_explainer, 'expected_value') else 0,
        data=valid_samples[:, top_features_idx],
        feature_names=top_feature_names
    )
    
    # Generate RTL Heatmap Plot using RTL Explainer
    print("\n Generating RTL Heatmap Plot...")
    print(" Each row = one sample, each column = one feature")
    print(" Red = High SHAP (pushes toward Abusive)")
    print(" Blue = Low SHAP (pushes toward Normal)")
    print(f" Showing TOP {max_features} features")
    
    # RTL explainer handles translation internally
    explainer_rtl.run_full_analysis(
        shap_values=shap_exp_heatmap,
        prediction_label=class_names[1],
        plot_fn=heatmap,
        max_features=max_features
    )
    plt.show()
    
    print(" Heatmap complete!")


```

```python
# ============================================================
# CELL 2b: CLASSIFICATION PERFORMANCE
# ============================================================
from sklearn.metrics import (confusion_matrix, classification_report,
                             f1_score, accuracy_score)

y_train_pred = model.predict(X_train_vec)
y_test_pred  = model.predict(X_test_vec)

tn, fp, fn, tp = confusion_matrix(y_test, y_test_pred).ravel()

print("="*60)
print("CONFUSION MATRIX (test set)")
print("="*60)
print(f"                  Predicted Normal   Predicted Abusive")
print(f"  Actual Normal        TN = {tn:<10} FP = {fp}")
print(f"  Actual Abusive       FN = {fn:<10} TP = {tp}")

print("\n" + "="*60)
print("METRICS")
print("="*60)
print(f"  Train accuracy : {accuracy_score(y_train, y_train_pred):.4f}")
print(f"  Test accuracy  : {accuracy_score(y_test,  y_test_pred):.4f}")
print(f"  Macro F1       : {f1_score(y_test, y_test_pred, average='macro'):.4f}")
print(f"  Weighted F1    : {f1_score(y_test, y_test_pred, average='weighted'):.4f}")
print(f"  F1 (Normal)    : {f1_score(y_test, y_test_pred, pos_label=0):.4f}")
print(f"  F1 (Abusive)   : {f1_score(y_test, y_test_pred, pos_label=1):.4f}")

print("\n" + "="*60)
print("PER-CLASS BREAKDOWN")
print("="*60)
print(classification_report(y_test, y_test_pred,
                            target_names=['Normal', 'Abusive'], digits=4))

# Majority-class baseline — the number that shows your model learned something
majority = max((y_test == 0).mean(), (y_test == 1).mean())
print(f"Majority-class baseline accuracy: {majority:.4f}")
print(f"Your model beats it by: {accuracy_score(y_test, y_test_pred) - majority:+.4f}")
```
